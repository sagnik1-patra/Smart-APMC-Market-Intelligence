```python
# ================================================================
# SMART APMC MARKET INTELLIGENCE
# COMPLETE FIXED ML + GRAPH GENERATION SCRIPT
#
# INPUT:
# C:\Users\sagni\Downloads\Smart APMC Market Intelligence\
# List_Of_APMC.csv
#
# OUTPUTS:
# cleaned_apmc_data.csv
# result.csv
# prediction.csv
# accuracy_graph.png
# accuracy_graph.svg
# comparison_graph.png
# comparison_graph.svg
# heatmap.png
# heatmap.svg
# result_graph.png
# result_graph.svg
# prediction_graph.png
# prediction_graph.svg
# confusion_matrix.png
# confusion_matrix.svg
# apmc_best_model.pkl
# ================================================================


# ================================================================
# 1. IMPORT LIBRARIES
# ================================================================

import os
import warnings
import pickle

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split

from sklearn.preprocessing import (
    LabelEncoder,
    StandardScaler,
    OneHotEncoder
)

from sklearn.compose import ColumnTransformer

from sklearn.pipeline import Pipeline

from sklearn.impute import SimpleImputer

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

from sklearn.linear_model import LogisticRegression

from sklearn.tree import DecisionTreeClassifier

from sklearn.ensemble import (
    RandomForestClassifier,
    ExtraTreesClassifier,
    GradientBoostingClassifier
)


warnings.filterwarnings("ignore")


# ================================================================
# 2. PATH CONFIGURATION
# ================================================================

BASE_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\Smart APMC Market Intelligence"
)


CSV_PATH = os.path.join(
    BASE_DIR,
    "List_Of_APMC.csv"
)


# ================================================================
# OUTPUT FILE PATHS
# ================================================================

CLEANED_DATA_PATH = os.path.join(
    BASE_DIR,
    "cleaned_apmc_data.csv"
)


RESULT_CSV_PATH = os.path.join(
    BASE_DIR,
    "result.csv"
)


PREDICTION_CSV_PATH = os.path.join(
    BASE_DIR,
    "prediction.csv"
)


MODEL_PATH = os.path.join(
    BASE_DIR,
    "apmc_best_model.pkl"
)


ACCURACY_PNG = os.path.join(
    BASE_DIR,
    "accuracy_graph.png"
)

ACCURACY_SVG = os.path.join(
    BASE_DIR,
    "accuracy_graph.svg"
)


COMPARISON_PNG = os.path.join(
    BASE_DIR,
    "comparison_graph.png"
)

COMPARISON_SVG = os.path.join(
    BASE_DIR,
    "comparison_graph.svg"
)


HEATMAP_PNG = os.path.join(
    BASE_DIR,
    "heatmap.png"
)

HEATMAP_SVG = os.path.join(
    BASE_DIR,
    "heatmap.svg"
)


RESULT_GRAPH_PNG = os.path.join(
    BASE_DIR,
    "result_graph.png"
)

RESULT_GRAPH_SVG = os.path.join(
    BASE_DIR,
    "result_graph.svg"
)


PREDICTION_GRAPH_PNG = os.path.join(
    BASE_DIR,
    "prediction_graph.png"
)

PREDICTION_GRAPH_SVG = os.path.join(
    BASE_DIR,
    "prediction_graph.svg"
)


CONFUSION_PNG = os.path.join(
    BASE_DIR,
    "confusion_matrix.png"
)

CONFUSION_SVG = os.path.join(
    BASE_DIR,
    "confusion_matrix.svg"
)


os.makedirs(
    BASE_DIR,
    exist_ok=True
)


# ================================================================
# 3. HEADER
# ================================================================

print("=" * 80)

print(
    "SMART APMC MARKET INTELLIGENCE"
)

print(
    "DATA CLEANING + MACHINE LEARNING + ANALYSIS + PREDICTION"
)

print("=" * 80)


# ================================================================
# 4. CHECK INPUT FILE
# ================================================================

if not os.path.exists(CSV_PATH):

    raise FileNotFoundError(
        "\nCSV file not found:\n"
        + CSV_PATH
    )


print(
    "\nInput CSV:"
)

print(
    CSV_PATH
)


# ================================================================
# 5. LOAD RAW CSV
# ================================================================

# IMPORTANT:
#
# header=None because this CSV contains several title/header rows.
#
# cp1252 works for the legacy Kannada text encoding contained
# in this particular file.


raw = pd.read_csv(

    CSV_PATH,

    encoding="cp1252",

    header=None,

    low_memory=False

)


print(
    "\nRaw dataset shape:",
    raw.shape
)


# ================================================================
# 6. KEEP THE REAL APMC TABLE
# ================================================================

# Columns 9-12 contain a separate summary table.
#
# Therefore only first 8 columns are used for the actual
# APMC / sub-market dataset.


raw = raw.iloc[
    :,
    0:8
].copy()


raw.columns = [

    "main_serial",

    "main_market_name",

    "main_est_year",

    "sub_serial",

    "sub_market_name",

    "sub_est_year",

    "area_serial",

    "market_area"

]


print(
    "\nAPMC table columns:"
)


for column in raw.columns:

    print(
        " -",
        column
    )


# ================================================================
# 7. CONVERT NUMERIC COLUMNS
# ================================================================

numeric_raw_columns = [

    "main_serial",

    "main_est_year",

    "sub_serial",

    "sub_est_year",

    "area_serial"

]


for column in numeric_raw_columns:

    raw[column] = pd.to_numeric(

        raw[column],

        errors="coerce"

    )


# ================================================================
# 8. CLEAN TEXT COLUMNS
# ================================================================

text_columns = [

    "main_market_name",

    "sub_market_name",

    "market_area"

]


for column in text_columns:

    raw[column] = (

        raw[column]

        .astype(str)

        .str.strip()

        .replace({

            "": np.nan,

            "nan": np.nan,

            "None": np.nan,

            "<NA>": np.nan

        })

    )


# ================================================================
# 9. REMOVE INVALID YEAR VALUES
# ================================================================

# Valid establishment years are assumed to lie between
# 1900 and 2030.


raw.loc[

    ~raw[
        "main_est_year"
    ].between(
        1900,
        2030
    ),

    "main_est_year"

] = np.nan


raw.loc[

    ~raw[
        "sub_est_year"
    ].between(
        1900,
        2030
    ),

    "sub_est_year"

] = np.nan


# ================================================================
# 10. RECONSTRUCT MAIN MARKET CONTEXT
# ================================================================

# The CSV doesn't repeat main-market details on every sub-market row.
#
# We therefore forward-fill the parent main-market context.


raw[
    "main_market_context"
] = (

    raw[
        "main_market_name"
    ]

    .where(

        raw[
            "main_est_year"
        ].notna()

    )

    .ffill()

)


raw[
    "main_year_context"
] = (

    raw[
        "main_est_year"
    ]

    .ffill()

)


# ================================================================
# 11. BUILD MAIN MARKET RECORDS
# ================================================================

main_rows = raw[

    raw[
        "main_est_year"
    ].notna()

    &

    raw[
        "main_market_name"
    ].notna()

].copy()


main_df = pd.DataFrame({

    "market_type":

        "Main Market",

    "market_name":

        main_rows[
            "main_market_name"
        ],

    "establishment_year":

        main_rows[
            "main_est_year"
        ],

    "parent_main_market":

        main_rows[
            "main_market_name"
        ],

    "parent_main_year":

        main_rows[
            "main_est_year"
        ],

    "market_area":

        main_rows[
            "market_area"
        ]

})


# ================================================================
# 12. BUILD SUB-MARKET RECORDS
# ================================================================

sub_rows = raw[

    raw[
        "sub_est_year"
    ].notna()

    &

    raw[
        "sub_market_name"
    ].notna()

].copy()


sub_df = pd.DataFrame({

    "market_type":

        "Sub Market",

    "market_name":

        sub_rows[
            "sub_market_name"
        ],

    "establishment_year":

        sub_rows[
            "sub_est_year"
        ],

    "parent_main_market":

        sub_rows[
            "main_market_context"
        ],

    "parent_main_year":

        sub_rows[
            "main_year_context"
        ],

    "market_area":

        sub_rows[
            "market_area"
        ]

})


# ================================================================
# 13. COMBINE MAIN + SUB MARKETS
# ================================================================

data = pd.concat(

    [
        main_df,
        sub_df
    ],

    ignore_index=True

)


# ================================================================
# 14. CLEAN NORMALIZED DATA
# ================================================================

data = data.dropna(

    subset=[

        "market_name",

        "establishment_year"

    ]

)


data.drop_duplicates(

    inplace=True

)


data.reset_index(

    drop=True,

    inplace=True

)


data[
    "record_id"
] = np.arange(

    1,

    len(data) + 1

)


# ================================================================
# 15. FORMAT YEAR COLUMNS
# ================================================================

data[
    "establishment_year"
] = (

    pd.to_numeric(

        data[
            "establishment_year"
        ],

        errors="coerce"

    )

    .astype(int)

)


data[
    "parent_main_year"
] = pd.to_numeric(

    data[
        "parent_main_year"
    ],

    errors="coerce"

)


# ================================================================
# 16. FEATURE ENGINEERING
# ================================================================

data[
    "market_name_length"
] = (

    data[
        "market_name"
    ]

    .astype(str)

    .str.len()

)


data[
    "parent_name_length"
] = (

    data[
        "parent_main_market"
    ]

    .fillna("")

    .astype(str)

    .str.len()

)


data[
    "area_name_length"
] = (

    data[
        "market_area"
    ]

    .fillna("")

    .astype(str)

    .str.len()

)


# ================================================================
# 17. CREATE MARKET-ERA TARGET
# ================================================================

# We cannot use columns 9-12 as a target.
#
# Instead create an interpretable classification target
# based on establishment period.


data[
    "market_era"
] = pd.cut(

    data[
        "establishment_year"
    ],

    bins=[

        1899,

        1990,

        2005,

        2030

    ],

    labels=[

        "Legacy (<=1990)",

        "Established (1991-2005)",

        "Modern (2006+)"

    ],

    include_lowest=True

)


data[
    "market_era"
] = (

    data[
        "market_era"
    ]

    .astype(str)

)


# ================================================================
# 18. REMOVE INVALID TARGETS
# ================================================================

data = data[

    data[
        "market_era"
    ]

    != "nan"

].copy()


data.reset_index(

    drop=True,

    inplace=True

)


# ================================================================
# 19. DISPLAY DISTRIBUTION
# ================================================================

print(
    "\nCleaned market records:",
    len(data)
)


print(
    "\nMarket era distribution:"
)


print(

    data[
        "market_era"
    ]

    .value_counts()

)


# ================================================================
# 20. SAVE CLEANED DATA
# ================================================================

data.to_csv(

    CLEANED_DATA_PATH,

    index=False,

    encoding="utf-8-sig"

)


print(
    "\nCleaned APMC data saved:"
)

print(
    CLEANED_DATA_PATH
)


# ================================================================
# 21. SELECT FEATURES
# ================================================================

# IMPORTANT:
#
# establishment_year is deliberately excluded.
#
# This prevents the model from directly reading the feature used
# to define market_era.


feature_columns = [

    "market_type",

    "parent_main_market",

    "parent_main_year",

    "market_area",

    "market_name_length",

    "parent_name_length",

    "area_name_length"

]


X = data[
    feature_columns
].copy()


y = data[
    "market_era"
].copy()


# ================================================================
# 22. RARE CLASS SAFEGUARD
# ================================================================

# This specifically fixes:
#
# ValueError:
# The least populated class in y has only 1 member


class_counts = (

    y.value_counts()

)


valid_classes = (

    class_counts[
        class_counts >= 2
    ]

    .index

)


valid_mask = y.isin(
    valid_classes
)


X = X[
    valid_mask
].copy()


y = y[
    valid_mask
].copy()


data_for_ml = data[
    valid_mask
].copy()


# ================================================================
# 23. LABEL ENCODING
# ================================================================

label_encoder = (
    LabelEncoder()
)


y_encoded = (
    label_encoder
    .fit_transform(
        y
    )
)


print(
    "\nTarget classes:"
)


for index, label in enumerate(

    label_encoder.classes_

):

    print(

        index,
        "->",
        label

    )


# ================================================================
# 24. NUMERIC / CATEGORICAL FEATURES
# ================================================================

numeric_features = [

    "parent_main_year",

    "market_name_length",

    "parent_name_length",

    "area_name_length"

]


categorical_features = [

    "market_type",

    "parent_main_market",

    "market_area"

]


# ================================================================
# 25. NUMERIC PIPELINE
# ================================================================

numeric_pipeline = Pipeline(

    steps=[

        (

            "imputer",

            SimpleImputer(
                strategy="median"
            )

        ),

        (

            "scaler",

            StandardScaler()

        )

    ]

)


# ================================================================
# 26. CATEGORICAL PIPELINE
# ================================================================

categorical_pipeline = Pipeline(

    steps=[

        (

            "imputer",

            SimpleImputer(
                strategy="most_frequent"
            )

        ),

        (

            "encoder",

            OneHotEncoder(

                handle_unknown="ignore",

                sparse_output=False

            )

        )

    ]

)


# ================================================================
# 27. PREPROCESSOR
# ================================================================

preprocessor = ColumnTransformer(

    transformers=[

        (

            "numeric",

            numeric_pipeline,

            numeric_features

        ),

        (

            "categorical",

            categorical_pipeline,

            categorical_features

        )

    ]

)


# ================================================================
# 28. SAFE TRAIN / TEST SPLIT
# ================================================================

class_counts_encoded = np.bincount(
    y_encoded
)


if np.min(
    class_counts_encoded
) >= 2:

    stratify_value = (
        y_encoded
    )

else:

    stratify_value = None


X_train, X_test, y_train, y_test = (
    train_test_split(

        X,

        y_encoded,

        test_size=0.25,

        random_state=42,

        stratify=stratify_value

    )
)


print(
    "\nTraining samples:",
    len(X_train)
)


print(
    "Testing samples:",
    len(X_test)
)


# ================================================================
# 29. DEFINE MODELS
# ================================================================

models = {

    "Logistic Regression":

        LogisticRegression(

            max_iter=3000,

            random_state=42

        ),


    "Decision Tree":

        DecisionTreeClassifier(

            max_depth=10,

            random_state=42

        ),


    "Random Forest":

        RandomForestClassifier(

            n_estimators=300,

            max_depth=15,

            random_state=42,

            n_jobs=-1

        ),


    "Extra Trees":

        ExtraTreesClassifier(

            n_estimators=300,

            random_state=42,

            n_jobs=-1

        ),


    "Gradient Boosting":

        GradientBoostingClassifier(

            n_estimators=150,

            random_state=42

        )

}


# ================================================================
# 30. TRAIN MODELS
# ================================================================

results = []

trained_models = {}


print(
    "\n"
    + "=" * 80
)

print(
    "MODEL TRAINING"
)

print(
    "=" * 80
)


for model_name, model in models.items():

    print(
        "\nTraining:",
        model_name
    )


    pipeline = Pipeline(

        steps=[

            (

                "preprocessor",

                preprocessor

            ),

            (

                "model",

                model

            )

        ]

    )


    pipeline.fit(

        X_train,

        y_train

    )


    predictions = pipeline.predict(
        X_test
    )


    accuracy = accuracy_score(

        y_test,

        predictions

    )


    precision = precision_score(

        y_test,

        predictions,

        average="weighted",

        zero_division=0

    )


    recall = recall_score(

        y_test,

        predictions,

        average="weighted",

        zero_division=0

    )


    f1 = f1_score(

        y_test,

        predictions,

        average="weighted",

        zero_division=0

    )


    results.append({

        "Model":

            model_name,

        "Accuracy":

            accuracy,

        "Precision":

            precision,

        "Recall":

            recall,

        "F1_Score":

            f1,

        "Accuracy_Percent":

            accuracy * 100,

        "Precision_Percent":

            precision * 100,

        "Recall_Percent":

            recall * 100,

        "F1_Percent":

            f1 * 100

    })


    trained_models[
        model_name
    ] = pipeline


    print(

        f"Accuracy  : "
        f"{accuracy * 100:.2f}%"

    )


    print(

        f"Precision : "
        f"{precision * 100:.2f}%"

    )


    print(

        f"Recall    : "
        f"{recall * 100:.2f}%"

    )


    print(

        f"F1 Score  : "
        f"{f1 * 100:.2f}%"

    )


# ================================================================
# 31. RESULT DATAFRAME
# ================================================================

result_df = pd.DataFrame(
    results
)


result_df = result_df.sort_values(

    by="Accuracy",

    ascending=False

)


result_df.reset_index(

    drop=True,

    inplace=True

)


# ================================================================
# 32. SAVE RESULT CSV
# ================================================================

result_df.to_csv(

    RESULT_CSV_PATH,

    index=False

)


print(
    "\nResult CSV saved:"
)

print(
    RESULT_CSV_PATH
)


print(
    "\nModel comparison:"
)

print(

    result_df[

        [

            "Model",

            "Accuracy_Percent",

            "Precision_Percent",

            "Recall_Percent",

            "F1_Percent"

        ]

    ]

)


# ================================================================
# 33. GET BEST MODEL
# ================================================================

best_model_name = (

    result_df
    .iloc[0][
        "Model"
    ]

)


best_accuracy = (

    result_df
    .iloc[0][
        "Accuracy"
    ]

)


best_model = (

    trained_models[
        best_model_name
    ]

)


print(
    "\nBest model:",
    best_model_name
)


print(

    "Best accuracy:",

    f"{best_accuracy * 100:.2f}%"

)


# ================================================================
# 34. SAVE BEST MODEL
# ================================================================

model_package = {

    "model":

        best_model,

    "label_encoder":

        label_encoder,

    "feature_columns":

        feature_columns,

    "target":

        "market_era",

    "best_model_name":

        best_model_name,

    "best_accuracy":

        float(
            best_accuracy
        )

}


with open(

    MODEL_PATH,

    "wb"

) as file:

    pickle.dump(

        model_package,

        file

    )


print(
    "\nBest model saved:"
)

print(
    MODEL_PATH
)


# ================================================================
# 35. ACCURACY GRAPH
# ================================================================

plt.figure(
    figsize=(11, 6)
)


plt.bar(

    result_df[
        "Model"
    ],

    result_df[
        "Accuracy_Percent"
    ]

)


plt.xlabel(
    "Machine Learning Model"
)


plt.ylabel(
    "Accuracy (%)"
)


plt.title(
    "Smart APMC Market Intelligence - Model Accuracy"
)


plt.ylim(
    0,
    100
)


plt.xticks(

    rotation=25,

    ha="right"

)


for i, value in enumerate(

    result_df[
        "Accuracy_Percent"
    ]

):

    plt.text(

        i,

        value + 1,

        f"{value:.2f}%",

        ha="center"

    )


plt.tight_layout()


plt.savefig(

    ACCURACY_PNG,

    dpi=300,

    bbox_inches="tight"

)


plt.savefig(

    ACCURACY_SVG,

    format="svg",

    bbox_inches="tight"

)


plt.show()

plt.close()


# ================================================================
# 36. COMPARISON GRAPH
# ================================================================

comparison_df = (

    result_df[

        [

            "Model",

            "Accuracy_Percent",

            "Precision_Percent",

            "Recall_Percent",

            "F1_Percent"

        ]

    ]

    .set_index(
        "Model"
    )

)


ax = comparison_df.plot(

    kind="bar",

    figsize=(13, 7)

)


plt.title(
    "Machine Learning Model Performance Comparison"
)


plt.xlabel(
    "Model"
)


plt.ylabel(
    "Performance (%)"
)


plt.ylim(
    0,
    100
)


plt.xticks(

    rotation=25,

    ha="right"

)


plt.legend([

    "Accuracy",

    "Precision",

    "Recall",

    "F1 Score"

])


plt.tight_layout()


plt.savefig(

    COMPARISON_PNG,

    dpi=300,

    bbox_inches="tight"

)


plt.savefig(

    COMPARISON_SVG,

    format="svg",

    bbox_inches="tight"

)


plt.show()

plt.close()


# ================================================================
# 37. HEATMAP DATA
# ================================================================

heatmap_df = data.copy()


# Encode selected categorical values for visualization


heatmap_df[
    "market_type_code"
] = (

    heatmap_df[
        "market_type"
    ]

    .map({

        "Main Market": 0,

        "Sub Market": 1

    })

)


heatmap_df[
    "market_era_code"
] = (

    heatmap_df[
        "market_era"
    ]

    .map({

        "Legacy (<=1990)": 0,

        "Established (1991-2005)": 1,

        "Modern (2006+)": 2

    })

)


heatmap_columns = [

    "establishment_year",

    "parent_main_year",

    "market_name_length",

    "parent_name_length",

    "area_name_length",

    "market_type_code",

    "market_era_code"

]


heatmap_numeric = (

    heatmap_df[
        heatmap_columns
    ]

    .apply(

        pd.to_numeric,

        errors="coerce"

    )

)


correlation_matrix = (
    heatmap_numeric.corr()
)


# ================================================================
# 38. HEATMAP
# ================================================================

plt.figure(
    figsize=(11, 8)
)


image = plt.imshow(

    correlation_matrix,

    cmap="coolwarm",

    aspect="auto",

    vmin=-1,

    vmax=1

)


plt.colorbar(

    image,

    label="Correlation"

)


plt.xticks(

    range(
        len(
            correlation_matrix.columns
        )
    ),

    correlation_matrix.columns,

    rotation=45,

    ha="right"

)


plt.yticks(

    range(
        len(
            correlation_matrix.columns
        )
    ),

    correlation_matrix.columns

)


for i in range(

    correlation_matrix.shape[0]

):

    for j in range(

        correlation_matrix.shape[1]

    ):

        value = (

            correlation_matrix
            .iloc[
                i,
                j
            ]

        )


        if not np.isnan(value):

            plt.text(

                j,

                i,

                f"{value:.2f}",

                ha="center",

                va="center",

                fontsize=9

            )


plt.title(
    "Smart APMC Feature Correlation Heatmap"
)


plt.tight_layout()


plt.savefig(

    HEATMAP_PNG,

    dpi=300,

    bbox_inches="tight"

)


plt.savefig(

    HEATMAP_SVG,

    format="svg",

    bbox_inches="tight"

)


plt.show()

plt.close()


# ================================================================
# 39. RESULT GRAPH - BEST MODEL
# ================================================================

best_row = result_df.iloc[0]


metric_names = [

    "Accuracy",

    "Precision",

    "Recall",

    "F1 Score"

]


metric_values = [

    best_row[
        "Accuracy_Percent"
    ],

    best_row[
        "Precision_Percent"
    ],

    best_row[
        "Recall_Percent"
    ],

    best_row[
        "F1_Percent"
    ]

]


plt.figure(
    figsize=(9, 6)
)


plt.bar(

    metric_names,

    metric_values

)


plt.title(

    f"Best Model Performance - "
    f"{best_model_name}"

)


plt.xlabel(
    "Metric"
)


plt.ylabel(
    "Score (%)"
)


plt.ylim(
    0,
    100
)


for i, value in enumerate(
    metric_values
):

    plt.text(

        i,

        value + 1,

        f"{value:.2f}%",

        ha="center"

    )


plt.tight_layout()


plt.savefig(

    RESULT_GRAPH_PNG,

    dpi=300,

    bbox_inches="tight"

)


plt.savefig(

    RESULT_GRAPH_SVG,

    format="svg",

    bbox_inches="tight"

)


plt.show()

plt.close()


# ================================================================
# 40. BEST MODEL PREDICTIONS
# ================================================================

best_predictions = (
    best_model.predict(
        X_test
    )
)


actual_labels = (

    label_encoder
    .inverse_transform(
        y_test
    )

)


predicted_labels = (

    label_encoder
    .inverse_transform(
        best_predictions
    )

)


# ================================================================
# 41. BUILD PREDICTION CSV
# ================================================================

prediction_df = pd.DataFrame({

    "Record_Number":

        np.arange(

            1,

            len(
                y_test
            ) + 1

        ),

    "Actual":

        actual_labels,

    "Predicted":

        predicted_labels

})


prediction_df[
    "Correct_Prediction"
] = np.where(

    prediction_df[
        "Actual"
    ]

    ==

    prediction_df[
        "Predicted"
    ],

    "Yes",

    "No"

)


# ================================================================
# 42. PREDICTION CONFIDENCE
# ================================================================

if hasattr(

    best_model,

    "predict_proba"

):

    probabilities = (

        best_model.predict_proba(
            X_test
        )

    )


    prediction_df[
        "Confidence_Percent"
    ] = (

        probabilities.max(
            axis=1
        )

        * 100

    )


else:

    prediction_df[
        "Confidence_Percent"
    ] = np.nan


prediction_df.to_csv(

    PREDICTION_CSV_PATH,

    index=False

)


print(
    "\nPrediction CSV saved:"
)

print(
    PREDICTION_CSV_PATH
)


print(
    "\nSample predictions:"
)


print(

    prediction_df.head(
        15
    )

)


# ================================================================
# 43. PREDICTION GRAPH
# ================================================================

actual_counts = (

    prediction_df[
        "Actual"
    ]

    .value_counts()

)


predicted_counts = (

    prediction_df[
        "Predicted"
    ]

    .value_counts()

)


classes = list(

    label_encoder.classes_

)


actual_values = [

    actual_counts.get(

        label,

        0

    )

    for label in classes

]


predicted_values = [

    predicted_counts.get(

        label,

        0

    )

    for label in classes

]


x = np.arange(
    len(classes)
)


width = 0.35


plt.figure(
    figsize=(11, 7)
)


plt.bar(

    x - width / 2,

    actual_values,

    width,

    label="Actual"

)


plt.bar(

    x + width / 2,

    predicted_values,

    width,

    label="Predicted"

)


plt.xlabel(
    "Market Era"
)


plt.ylabel(
    "Number of Records"
)


plt.title(

    f"Actual vs Predicted Market Era - "
    f"{best_model_name}"

)


plt.xticks(

    x,

    classes,

    rotation=20,

    ha="right"

)


plt.legend()


plt.tight_layout()


plt.savefig(

    PREDICTION_GRAPH_PNG,

    dpi=300,

    bbox_inches="tight"

)


plt.savefig(

    PREDICTION_GRAPH_SVG,

    format="svg",

    bbox_inches="tight"

)


plt.show()

plt.close()


# ================================================================
# 44. CONFUSION MATRIX
# ================================================================

cm = confusion_matrix(

    y_test,

    best_predictions

)


plt.figure(
    figsize=(9, 7)
)


cm_image = plt.imshow(

    cm,

    cmap="Blues"

)


plt.colorbar(
    cm_image
)


plt.xticks(

    range(
        len(classes)
    ),

    classes,

    rotation=25,

    ha="right"

)


plt.yticks(

    range(
        len(classes)
    ),

    classes

)


for i in range(
    cm.shape[0]
):

    for j in range(
        cm.shape[1]
    ):

        plt.text(

            j,

            i,

            cm[i, j],

            ha="center",

            va="center"

        )


plt.xlabel(
    "Predicted Class"
)


plt.ylabel(
    "Actual Class"
)


plt.title(

    f"Confusion Matrix - "
    f"{best_model_name}"

)


plt.tight_layout()


plt.savefig(

    CONFUSION_PNG,

    dpi=300,

    bbox_inches="tight"

)


plt.savefig(

    CONFUSION_SVG,

    format="svg",

    bbox_inches="tight"

)


plt.show()

plt.close()


# ================================================================
# 45. FILE VERIFICATION
# ================================================================

generated_files = [

    CLEANED_DATA_PATH,

    RESULT_CSV_PATH,

    PREDICTION_CSV_PATH,

    MODEL_PATH,

    ACCURACY_PNG,

    ACCURACY_SVG,

    COMPARISON_PNG,

    COMPARISON_SVG,

    HEATMAP_PNG,

    HEATMAP_SVG,

    RESULT_GRAPH_PNG,

    RESULT_GRAPH_SVG,

    PREDICTION_GRAPH_PNG,

    PREDICTION_GRAPH_SVG,

    CONFUSION_PNG,

    CONFUSION_SVG

]


print(
    "\n"
    + "=" * 80
)

print(
    "GENERATED FILES"
)

print(
    "=" * 80
)


for i, file_path in enumerate(

    generated_files,

    start=1

):

    if os.path.exists(
        file_path
    ):

        size_kb = (

            os.path.getsize(
                file_path
            )

            / 1024

        )


        print(

            f"{i}. "
            f"{os.path.basename(file_path)} "
            f"- {size_kb:.2f} KB"

        )


    else:

        print(

            f"{i}. "
            f"{os.path.basename(file_path)} "
            "- FAILED"

        )


# ================================================================
# 46. FINAL SUMMARY
# ================================================================

print(
    "\n"
    + "=" * 80
)

print(
    "SMART APMC MARKET INTELLIGENCE COMPLETED"
)

print(
    "=" * 80
)


print(
    "\nTotal usable market records:",
    len(data)
)


print(
    "\nBest model:",
    best_model_name
)


print(

    "Best accuracy:",

    f"{best_accuracy * 100:.2f}%"

)


print(
    "\nAll outputs saved inside:"
)


print(
    BASE_DIR
)


print(
    "\nDONE!"
)
```
