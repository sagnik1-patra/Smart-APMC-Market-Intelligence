```python
import os
import warnings
import pickle

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt

from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    cross_val_score
)

from sklearn.preprocessing import (
    LabelEncoder,
    StandardScaler,
    OneHotEncoder
)

from sklearn.compose import ColumnTransformer

from sklearn.pipeline import Pipeline

from sklearn.impute import SimpleImputer

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

from sklearn.linear_model import LogisticRegression

from sklearn.tree import DecisionTreeClassifier

from sklearn.ensemble import (
    RandomForestClassifier,
    ExtraTreesClassifier,
    GradientBoostingClassifier
)


warnings.filterwarnings("ignore")


# ======================================================================
# 2. RANDOM SEED
# ======================================================================

RANDOM_STATE = 42

np.random.seed(
    RANDOM_STATE
)


# ======================================================================
# 3. PROJECT PATH
# ======================================================================

BASE_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\Smart APMC Market Intelligence"
)


CSV_PATH = os.path.join(
    BASE_DIR,
    "List_Of_APMC.csv"
)


os.makedirs(
    BASE_DIR,
    exist_ok=True
)


# ======================================================================
# 4. OUTPUT PATHS
# ======================================================================

CLEANED_DATA_PATH = os.path.join(
    BASE_DIR,
    "pso_cleaned_apmc_data.csv"
)


SELECTED_FEATURES_PATH = os.path.join(
    BASE_DIR,
    "pso_selected_features.csv"
)


FITNESS_HISTORY_PATH = os.path.join(
    BASE_DIR,
    "pso_fitness_history.csv"
)


RESULT_CSV_PATH = os.path.join(
    BASE_DIR,
    "pso_result.csv"
)


PREDICTION_CSV_PATH = os.path.join(
    BASE_DIR,
    "pso_prediction.csv"
)


MODEL_PATH = os.path.join(
    BASE_DIR,
    "pso_best_model.pkl"
)


# Accuracy
ACCURACY_PNG = os.path.join(
    BASE_DIR,
    "pso_accuracy_graph.png"
)

ACCURACY_SVG = os.path.join(
    BASE_DIR,
    "pso_accuracy_graph.svg"
)


# Comparison
COMPARISON_PNG = os.path.join(
    BASE_DIR,
    "pso_comparison_graph.png"
)

COMPARISON_SVG = os.path.join(
    BASE_DIR,
    "pso_comparison_graph.svg"
)


# Heatmap
HEATMAP_PNG = os.path.join(
    BASE_DIR,
    "pso_heatmap.png"
)

HEATMAP_SVG = os.path.join(
    BASE_DIR,
    "pso_heatmap.svg"
)


# Result graph
RESULT_GRAPH_PNG = os.path.join(
    BASE_DIR,
    "pso_result_graph.png"
)

RESULT_GRAPH_SVG = os.path.join(
    BASE_DIR,
    "pso_result_graph.svg"
)


# Prediction graph
PREDICTION_GRAPH_PNG = os.path.join(
    BASE_DIR,
    "pso_prediction_graph.png"
)

PREDICTION_GRAPH_SVG = os.path.join(
    BASE_DIR,
    "pso_prediction_graph.svg"
)


# Confusion matrix
CONFUSION_PNG = os.path.join(
    BASE_DIR,
    "pso_confusion_matrix.png"
)

CONFUSION_SVG = os.path.join(
    BASE_DIR,
    "pso_confusion_matrix.svg"
)


# PSO convergence
CONVERGENCE_PNG = os.path.join(
    BASE_DIR,
    "pso_convergence_graph.png"
)

CONVERGENCE_SVG = os.path.join(
    BASE_DIR,
    "pso_convergence_graph.svg"
)


# Feature selection graph
FEATURE_GRAPH_PNG = os.path.join(
    BASE_DIR,
    "pso_feature_selection_graph.png"
)

FEATURE_GRAPH_SVG = os.path.join(
    BASE_DIR,
    "pso_feature_selection_graph.svg"
)


# ======================================================================
# 5. HEADER
# ======================================================================

print(
    "=" * 80
)

print(
    "SMART APMC MARKET INTELLIGENCE"
)

print(
    "PARTICLE SWARM OPTIMIZATION + MACHINE LEARNING"
)

print(
    "=" * 80
)


# ======================================================================
# 6. CHECK CSV
# ======================================================================

if not os.path.exists(
    CSV_PATH
):

    raise FileNotFoundError(
        "\nCSV file not found:\n"
        + CSV_PATH
    )


print(
    "\nInput dataset:"
)

print(
    CSV_PATH
)


# ======================================================================
# 7. LOAD DATASET
# ======================================================================

raw = pd.read_csv(

    CSV_PATH,

    encoding="cp1252",

    header=None,

    low_memory=False

)


print(
    "\nRaw dataset shape:",
    raw.shape
)


# ======================================================================
# 8. KEEP ONLY REAL APMC TABLE
# ======================================================================

# The rightmost columns contain an independent summary table.
# Therefore columns 0 to 7 are used.

raw = raw.iloc[
    :,
    0:8
].copy()


raw.columns = [

    "main_serial",

    "main_market_name",

    "main_est_year",

    "sub_serial",

    "sub_market_name",

    "sub_est_year",

    "area_serial",

    "market_area"

]


print(
    "\nColumns extracted:"
)


for column in raw.columns:

    print(
        " -",
        column
    )


# ======================================================================
# 9. NUMERIC CONVERSION
# ======================================================================

numeric_raw_columns = [

    "main_serial",

    "main_est_year",

    "sub_serial",

    "sub_est_year",

    "area_serial"

]


for column in numeric_raw_columns:

    raw[column] = pd.to_numeric(

        raw[column],

        errors="coerce"

    )


# ======================================================================
# 10. TEXT CLEANING
# ======================================================================

text_columns = [

    "main_market_name",

    "sub_market_name",

    "market_area"

]


for column in text_columns:

    raw[column] = (

        raw[column]

        .astype(str)

        .str.strip()

        .replace({

            "": np.nan,

            "nan": np.nan,

            "None": np.nan,

            "<NA>": np.nan

        })

    )


# ======================================================================
# 11. REMOVE INVALID YEARS
# ======================================================================

raw.loc[

    ~raw[
        "main_est_year"
    ].between(
        1900,
        2030
    ),

    "main_est_year"

] = np.nan


raw.loc[

    ~raw[
        "sub_est_year"
    ].between(
        1900,
        2030
    ),

    "sub_est_year"

] = np.nan


# ======================================================================
# 12. FORWARD-FILL PARENT APMC INFORMATION
# ======================================================================

raw[
    "main_market_context"
] = (

    raw[
        "main_market_name"
    ]

    .where(

        raw[
            "main_est_year"
        ].notna()

    )

    .ffill()

)


raw[
    "main_year_context"
] = (

    raw[
        "main_est_year"
    ]

    .ffill()

)


# ======================================================================
# 13. CREATE MAIN MARKET DATASET
# ======================================================================

main_rows = raw[

    raw[
        "main_est_year"
    ].notna()

    &

    raw[
        "main_market_name"
    ].notna()

].copy()


main_df = pd.DataFrame({

    "market_type":

        "Main Market",

    "market_name":

        main_rows[
            "main_market_name"
        ],

    "establishment_year":

        main_rows[
            "main_est_year"
        ],

    "parent_main_market":

        main_rows[
            "main_market_name"
        ],

    "parent_main_year":

        main_rows[
            "main_est_year"
        ],

    "market_area":

        main_rows[
            "market_area"
        ]

})


# ======================================================================
# 14. CREATE SUB-MARKET DATASET
# ======================================================================

sub_rows = raw[

    raw[
        "sub_est_year"
    ].notna()

    &

    raw[
        "sub_market_name"
    ].notna()

].copy()


sub_df = pd.DataFrame({

    "market_type":

        "Sub Market",

    "market_name":

        sub_rows[
            "sub_market_name"
        ],

    "establishment_year":

        sub_rows[
            "sub_est_year"
        ],

    "parent_main_market":

        sub_rows[
            "main_market_context"
        ],

    "parent_main_year":

        sub_rows[
            "main_year_context"
        ],

    "market_area":

        sub_rows[
            "market_area"
        ]

})


# ======================================================================
# 15. COMBINE DATA
# ======================================================================

data = pd.concat(

    [
        main_df,
        sub_df
    ],

    ignore_index=True

)


data.dropna(

    subset=[

        "market_name",

        "establishment_year"

    ],

    inplace=True

)


data.drop_duplicates(
    inplace=True
)


data.reset_index(
    drop=True,
    inplace=True
)


# ======================================================================
# 16. ADD RECORD ID
# ======================================================================

data[
    "record_id"
] = np.arange(

    1,

    len(data) + 1

)


# ======================================================================
# 17. YEAR CLEANING
# ======================================================================

data[
    "establishment_year"
] = (

    pd.to_numeric(

        data[
            "establishment_year"
        ],

        errors="coerce"

    )

    .astype(int)

)


data[
    "parent_main_year"
] = pd.to_numeric(

    data[
        "parent_main_year"
    ],

    errors="coerce"

)


# ======================================================================
# 18. FEATURE ENGINEERING
# ======================================================================

data[
    "market_name_length"
] = (

    data[
        "market_name"
    ]

    .astype(str)

    .str.len()

)


data[
    "parent_name_length"
] = (

    data[
        "parent_main_market"
    ]

    .fillna("")

    .astype(str)

    .str.len()

)


data[
    "area_name_length"
] = (

    data[
        "market_area"
    ]

    .fillna("")

    .astype(str)

    .str.len()

)


# ======================================================================
# 19. TARGET VARIABLE
# ======================================================================

data[
    "market_era"
] = pd.cut(

    data[
        "establishment_year"
    ],

    bins=[

        1899,

        1990,

        2005,

        2030

    ],

    labels=[

        "Legacy (<=1990)",

        "Established (1991-2005)",

        "Modern (2006+)"

    ],

    include_lowest=True

)


data[
    "market_era"
] = (

    data[
        "market_era"
    ]

    .astype(str)

)


data = data[

    data[
        "market_era"
    ] != "nan"

].copy()


data.reset_index(
    drop=True,
    inplace=True
)


# ======================================================================
# 20. SAVE CLEANED PSO DATASET
# ======================================================================

data.to_csv(

    CLEANED_DATA_PATH,

    index=False,

    encoding="utf-8-sig"

)


print(
    "\nCleaned records:",
    len(data)
)


print(
    "\nClass distribution:"
)


print(

    data[
        "market_era"
    ]

    .value_counts()

)


print(
    "\nSaved:"
)

print(
    CLEANED_DATA_PATH
)


# ======================================================================
# 21. ALL CANDIDATE FEATURES
# ======================================================================

# establishment_year is excluded because market_era was derived
# directly from establishment_year.
#
# Including it would cause target leakage.

all_features = [

    "market_type",

    "parent_main_market",

    "parent_main_year",

    "market_area",

    "market_name_length",

    "parent_name_length",

    "area_name_length"

]


numeric_all = [

    "parent_main_year",

    "market_name_length",

    "parent_name_length",

    "area_name_length"

]


categorical_all = [

    "market_type",

    "parent_main_market",

    "market_area"

]


X = data[
    all_features
].copy()


y = data[
    "market_era"
].copy()


# ======================================================================
# 22. REMOVE RARE CLASSES
# ======================================================================

class_counts = y.value_counts()


valid_classes = (

    class_counts[
        class_counts >= 2
    ]

    .index

)


valid_mask = y.isin(
    valid_classes
)


X = X[
    valid_mask
].copy()


y = y[
    valid_mask
].copy()


X.reset_index(
    drop=True,
    inplace=True
)


y.reset_index(
    drop=True,
    inplace=True
)


# ======================================================================
# 23. LABEL ENCODING
# ======================================================================

label_encoder = LabelEncoder()


y_encoded = (

    label_encoder

    .fit_transform(
        y
    )

)


print(
    "\nTarget classes:"
)


for i, class_name in enumerate(
    label_encoder.classes_
):

    print(
        i,
        "->",
        class_name
    )


# ======================================================================
# 24. TRAIN TEST SPLIT BEFORE PSO
# ======================================================================

encoded_counts = np.bincount(
    y_encoded
)


if np.min(
    encoded_counts
) >= 2:

    stratify_value = y_encoded

else:

    stratify_value = None


X_train, X_test, y_train, y_test = (
    train_test_split(

        X,

        y_encoded,

        test_size=0.25,

        random_state=RANDOM_STATE,

        stratify=stratify_value

    )
)


print(
    "\nTraining records:",
    len(X_train)
)


print(
    "Testing records:",
    len(X_test)
)


# ======================================================================
# 25. FUNCTION TO CREATE PREPROCESSOR
# ======================================================================

def create_preprocessor(selected_features):

    selected_numeric = [

        feature

        for feature in numeric_all

        if feature in selected_features

    ]


    selected_categorical = [

        feature

        for feature in categorical_all

        if feature in selected_features

    ]


    transformers = []


    # --------------------------------------------------------------
    # Numeric pipeline
    # --------------------------------------------------------------

    if len(
        selected_numeric
    ) > 0:

        numeric_pipeline = Pipeline(

            steps=[

                (

                    "imputer",

                    SimpleImputer(
                        strategy="median"
                    )

                ),

                (

                    "scaler",

                    StandardScaler()

                )

            ]

        )


        transformers.append(

            (

                "numeric",

                numeric_pipeline,

                selected_numeric

            )

        )


    # --------------------------------------------------------------
    # Categorical pipeline
    # --------------------------------------------------------------

    if len(
        selected_categorical
    ) > 0:

        categorical_pipeline = Pipeline(

            steps=[

                (

                    "imputer",

                    SimpleImputer(
                        strategy="most_frequent"
                    )

                ),

                (

                    "encoder",

                    OneHotEncoder(

                        handle_unknown="ignore",

                        sparse_output=False

                    )

                )

            ]

        )


        transformers.append(

            (

                "categorical",

                categorical_pipeline,

                selected_categorical

            )

        )


    return ColumnTransformer(

        transformers=transformers

    )


# ======================================================================
# 26. PSO CONFIGURATION
# ======================================================================

NUMBER_OF_PARTICLES = 15

NUMBER_OF_ITERATIONS = 20


# PSO coefficients

INERTIA_WEIGHT = 0.7

COGNITIVE_COEFFICIENT = 1.5

SOCIAL_COEFFICIENT = 1.5


# Penalty encourages smaller feature subsets.

FEATURE_PENALTY = 0.02


number_of_features = len(
    all_features
)


print(
    "\n"
    + "=" * 80
)

print(
    "PSO FEATURE SELECTION"
)

print(
    "=" * 80
)


print(
    "\nNumber of candidate features:",
    number_of_features
)


print(
    "Particles:",
    NUMBER_OF_PARTICLES
)


print(
    "Iterations:",
    NUMBER_OF_ITERATIONS
)


# ======================================================================
# 27. CV CONFIGURATION
# ======================================================================

minimum_class_size = np.min(
    np.bincount(
        y_train
    )
)


cv_splits = min(
    3,
    minimum_class_size
)


if cv_splits < 2:

    cv_splits = 2


cv = StratifiedKFold(

    n_splits=cv_splits,

    shuffle=True,

    random_state=RANDOM_STATE

)


# ======================================================================
# 28. PSO FITNESS FUNCTION
# ======================================================================

fitness_cache = {}


def evaluate_particle(
    binary_position
):

    # --------------------------------------------------------------
    # Convert binary vector into feature names
    # --------------------------------------------------------------

    selected_indices = np.where(
        binary_position == 1
    )[0]


    # Never allow zero selected features.

    if len(
        selected_indices
    ) == 0:

        return 1.0


    selected_features = [

        all_features[index]

        for index in selected_indices

    ]


    # --------------------------------------------------------------
    # Cache identical feature combinations
    # --------------------------------------------------------------

    cache_key = tuple(
        binary_position.tolist()
    )


    if cache_key in fitness_cache:

        return fitness_cache[
            cache_key
        ]


    # --------------------------------------------------------------
    # Build preprocessing pipeline
    # --------------------------------------------------------------

    preprocessor = create_preprocessor(
        selected_features
    )


    model = RandomForestClassifier(

        n_estimators=120,

        max_depth=12,

        random_state=RANDOM_STATE,

        n_jobs=-1

    )


    pipeline = Pipeline(

        steps=[

            (
                "preprocessor",
                preprocessor
            ),

            (
                "model",
                model
            )

        ]

    )


    try:

        scores = cross_val_score(

            pipeline,

            X_train[
                selected_features
            ],

            y_train,

            cv=cv,

            scoring="accuracy",

            n_jobs=-1

        )


        cv_accuracy = np.mean(
            scores
        )


    except Exception:

        cv_accuracy = 0


    # --------------------------------------------------------------
    # Feature selection penalty
    # --------------------------------------------------------------

    feature_ratio = (

        len(
            selected_features
        )

        /

        number_of_features

    )


    # PSO minimizes fitness:
    #
    # lower fitness = better
    #
    # fitness = classification error + feature penalty

    fitness = (

        (1 - cv_accuracy)

        +

        FEATURE_PENALTY
        *
        feature_ratio

    )


    fitness_cache[
        cache_key
    ] = fitness


    return fitness


# ======================================================================
# 29. INITIALIZE PSO PARTICLES
# ======================================================================

positions = np.random.uniform(

    low=0,

    high=1,

    size=(

        NUMBER_OF_PARTICLES,

        number_of_features

    )

)


velocities = np.random.uniform(

    low=-1,

    high=1,

    size=(

        NUMBER_OF_PARTICLES,

        number_of_features

    )

)


# Convert continuous probabilities into binary features

binary_positions = (

    positions >= 0.5

).astype(int)


# Make sure every particle has at least one feature.

for particle_index in range(
    NUMBER_OF_PARTICLES
):

    if np.sum(
        binary_positions[
            particle_index
        ]
    ) == 0:

        random_feature = np.random.randint(
            number_of_features
        )

        binary_positions[
            particle_index,
            random_feature
        ] = 1


# ======================================================================
# 30. INITIAL FITNESS
# ======================================================================

personal_best_positions = (
    binary_positions.copy()
)


personal_best_fitness = np.array([

    evaluate_particle(
        particle
    )

    for particle in binary_positions

])


best_particle_index = np.argmin(
    personal_best_fitness
)


global_best_position = (

    personal_best_positions[
        best_particle_index
    ].copy()

)


global_best_fitness = (

    personal_best_fitness[
        best_particle_index
    ]

)


fitness_history = []


# ======================================================================
# 31. PSO OPTIMIZATION LOOP
# ======================================================================

for iteration in range(
    NUMBER_OF_ITERATIONS
):

    for particle_index in range(
        NUMBER_OF_PARTICLES
    ):

        # ----------------------------------------------------------
        # Random PSO factors
        # ----------------------------------------------------------

        r1 = np.random.random(
            number_of_features
        )

        r2 = np.random.random(
            number_of_features
        )


        # ----------------------------------------------------------
        # Velocity update
        # ----------------------------------------------------------

        cognitive_component = (

            COGNITIVE_COEFFICIENT

            *

            r1

            *

            (

                personal_best_positions[
                    particle_index
                ]

                -

                positions[
                    particle_index
                ]

            )

        )


        social_component = (

            SOCIAL_COEFFICIENT

            *

            r2

            *

            (

                global_best_position

                -

                positions[
                    particle_index
                ]

            )

        )


        velocities[
            particle_index
        ] = (

            INERTIA_WEIGHT

            *

            velocities[
                particle_index
            ]

            +

            cognitive_component

            +

            social_component

        )


        # ----------------------------------------------------------
        # Binary PSO sigmoid
        # ----------------------------------------------------------

        sigmoid_probability = (

            1

            /

            (

                1

                +

                np.exp(

                    -velocities[
                        particle_index
                    ]

                )

            )

        )


        random_values = np.random.random(
            number_of_features
        )


        new_binary_position = (

            random_values

            <

            sigmoid_probability

        ).astype(int)


        # At least one feature must remain selected.

        if np.sum(
            new_binary_position
        ) == 0:

            new_binary_position[
                np.random.randint(
                    number_of_features
                )
            ] = 1


        binary_positions[
            particle_index
        ] = new_binary_position


        positions[
            particle_index
        ] = sigmoid_probability


        # ----------------------------------------------------------
        # Evaluate new particle
        # ----------------------------------------------------------

        current_fitness = evaluate_particle(
            new_binary_position
        )


        # ----------------------------------------------------------
        # Personal best update
        # ----------------------------------------------------------

        if (
            current_fitness

            <

            personal_best_fitness[
                particle_index
            ]
        ):

            personal_best_fitness[
                particle_index
            ] = current_fitness


            personal_best_positions[
                particle_index
            ] = (

                new_binary_position.copy()

            )


        # ----------------------------------------------------------
        # Global best update
        # ----------------------------------------------------------

        if (
            current_fitness

            <

            global_best_fitness
        ):

            global_best_fitness = (
                current_fitness
            )


            global_best_position = (

                new_binary_position.copy()

            )


    # --------------------------------------------------------------
    # Store convergence
    # --------------------------------------------------------------

    fitness_history.append(
        global_best_fitness
    )


    selected_count = np.sum(
        global_best_position
    )


    print(

        f"Iteration "
        f"{iteration + 1:02d}/"
        f"{NUMBER_OF_ITERATIONS}"

        " | "

        f"Best Fitness: "
        f"{global_best_fitness:.6f}"

        " | "

        f"Selected Features: "
        f"{selected_count}"

    )


# ======================================================================
# 32. PSO SELECTED FEATURES
# ======================================================================

selected_indices = np.where(
    global_best_position == 1
)[0]


selected_features = [

    all_features[index]

    for index in selected_indices

]


print(
    "\n"
    + "=" * 80
)


print(
    "PSO OPTIMIZATION COMPLETED"
)


print(
    "=" * 80
)


print(
    "\nSelected features:"
)


for feature in selected_features:

    print(
        " -",
        feature
    )


print(
    "\nTotal selected:",
    len(selected_features)
)


print(
    "Original features:",
    len(all_features)
)


# ======================================================================
# 33. SAVE SELECTED FEATURES CSV
# ======================================================================

selected_feature_df = pd.DataFrame({

    "Feature":
        all_features,

    "Selected":
        [

            "Yes"

            if feature in selected_features

            else "No"

            for feature in all_features

        ],

    "PSO_Binary_Value":
        global_best_position.astype(int)

})


selected_feature_df.to_csv(

    SELECTED_FEATURES_PATH,

    index=False

)


print(
    "\nSelected features saved:"
)

print(
    SELECTED_FEATURES_PATH
)


# ======================================================================
# 34. SAVE FITNESS HISTORY CSV
# ======================================================================

fitness_df = pd.DataFrame({

    "Iteration":

        np.arange(

            1,

            len(
                fitness_history
            ) + 1

        ),

    "Best_Fitness":

        fitness_history

})


fitness_df[
    "Estimated_Optimization_Score"
] = (

    1

    -

    fitness_df[
        "Best_Fitness"
    ]

)


fitness_df.to_csv(

    FITNESS_HISTORY_PATH,

    index=False

)


print(
    "\nFitness history saved:"
)

print(
    FITNESS_HISTORY_PATH
)


# ======================================================================
# 35. PSO CONVERGENCE GRAPH
# ======================================================================

plt.figure(
    figsize=(10, 6)
)


plt.plot(

    range(
        1,
        len(
            fitness_history
        ) + 1
    ),

    fitness_history,

    marker="o"

)


plt.xlabel(
    "PSO Iteration"
)


plt.ylabel(
    "Best Fitness"
)


plt.title(
    "PSO Convergence - Smart APMC Market Intelligence"
)


plt.grid(
    alpha=0.3
)


plt.tight_layout()


plt.savefig(

    CONVERGENCE_PNG,

    dpi=300,

    bbox_inches="tight"

)


plt.savefig(

    CONVERGENCE_SVG,

    format="svg",

    bbox_inches="tight"

)


plt.show()

plt.close()


# ======================================================================
# 36. FEATURE SELECTION GRAPH
# ======================================================================

plt.figure(
    figsize=(11, 6)
)


plt.bar(

    all_features,

    global_best_position

)


plt.xlabel(
    "Candidate Features"
)


plt.ylabel(
    "PSO Selection"
)


plt.title(
    "PSO Feature Selection"
)


plt.yticks(
    [0, 1],
    [
        "Not Selected",
        "Selected"
    ]
)


plt.xticks(

    rotation=35,

    ha="right"

)


plt.tight_layout()


plt.savefig(

    FEATURE_GRAPH_PNG,

    dpi=300,

    bbox_inches="tight"

)


plt.savefig(

    FEATURE_GRAPH_SVG,

    format="svg",

    bbox_inches="tight"

)


plt.show()

plt.close()


# ======================================================================
# 37. BUILD FINAL PSO PREPROCESSOR
# ======================================================================

final_preprocessor = create_preprocessor(
    selected_features
)


# ======================================================================
# 38. DEFINE FINAL MODELS
# ======================================================================

models = {

    "Logistic Regression":

        LogisticRegression(

            max_iter=3000,

            random_state=RANDOM_STATE

        ),


    "Decision Tree":

        DecisionTreeClassifier(

            max_depth=10,

            random_state=RANDOM_STATE

        ),


    "Random Forest":

        RandomForestClassifier(

            n_estimators=300,

            max_depth=15,

            random_state=RANDOM_STATE,

            n_jobs=-1

        ),


    "Extra Trees":

        ExtraTreesClassifier(

            n_estimators=300,

            random_state=RANDOM_STATE,

            n_jobs=-1

        ),


    "Gradient Boosting":

        GradientBoostingClassifier(

            n_estimators=150,

            learning_rate=0.1,

            random_state=RANDOM_STATE

        )

}


# ======================================================================
# 39. TRAIN FINAL PSO-OPTIMIZED MODELS
# ======================================================================

results = []

trained_models = {}


print(
    "\n"
    + "=" * 80
)


print(
    "PSO-OPTIMIZED MODEL TRAINING"
)


print(
    "=" * 80
)


for model_name, model in models.items():

    print(
        "\nTraining:",
        model_name
    )


    pipeline = Pipeline(

        steps=[

            (
                "preprocessor",
                final_preprocessor
            ),

            (
                "model",
                model
            )

        ]

    )


    try:

        pipeline.fit(

            X_train[
                selected_features
            ],

            y_train

        )


        predictions = pipeline.predict(

            X_test[
                selected_features
            ]

        )


        accuracy = accuracy_score(

            y_test,

            predictions

        )


        precision = precision_score(

            y_test,

            predictions,

            average="weighted",

            zero_division=0

        )


        recall = recall_score(

            y_test,

            predictions,

            average="weighted",

            zero_division=0

        )


        f1 = f1_score(

            y_test,

            predictions,

            average="weighted",

            zero_division=0

        )


        results.append({

            "Optimization":

                "PSO",

            "Model":

                model_name,

            "Selected_Features":

                len(
                    selected_features
                ),

            "Accuracy":

                accuracy,

            "Precision":

                precision,

            "Recall":

                recall,

            "F1_Score":

                f1,

            "Accuracy_Percent":

                accuracy * 100,

            "Precision_Percent":

                precision * 100,

            "Recall_Percent":

                recall * 100,

            "F1_Percent":

                f1 * 100

        })


        trained_models[
            model_name
        ] = pipeline


        print(
            f"Accuracy  : "
            f"{accuracy * 100:.2f}%"
        )


        print(
            f"Precision : "
            f"{precision * 100:.2f}%"
        )


        print(
            f"Recall    : "
            f"{recall * 100:.2f}%"
        )


        print(
            f"F1 Score  : "
            f"{f1 * 100:.2f}%"
        )


    except Exception as error:

        print(
            "Model failed:",
            error
        )


# ======================================================================
# 40. CREATE RESULT DATAFRAME
# ======================================================================

if len(
    results
) == 0:

    raise RuntimeError(
        "No machine learning model completed successfully."
    )


result_df = pd.DataFrame(
    results
)


result_df = result_df.sort_values(

    by="Accuracy",

    ascending=False

)


result_df.reset_index(
    drop=True,
    inplace=True
)


# ======================================================================
# 41. SAVE PSO RESULT CSV
# ======================================================================

result_df.to_csv(

    RESULT_CSV_PATH,

    index=False

)


print(
    "\nPSO result CSV saved:"
)

print(
    RESULT_CSV_PATH
)


print(
    "\nModel comparison:"
)


print(

    result_df[

        [

            "Model",

            "Accuracy_Percent",

            "Precision_Percent",

            "Recall_Percent",

            "F1_Percent"

        ]

    ]

)


# ======================================================================
# 42. FIND BEST MODEL
# ======================================================================

best_model_name = (

    result_df
    .iloc[0][
        "Model"
    ]

)


best_accuracy = (

    result_df
    .iloc[0][
        "Accuracy"
    ]

)


best_model = trained_models[
    best_model_name
]


print(
    "\nPSO Best Model:",
    best_model_name
)


print(

    "PSO Best Accuracy:",

    f"{best_accuracy * 100:.2f}%"

)


# ======================================================================
# 43. SAVE PSO BEST MODEL
# ======================================================================

model_package = {

    "optimization_algorithm":

        "Particle Swarm Optimization",

    "optimization_type":

        "Binary PSO Feature Selection",

    "model":

        best_model,

    "label_encoder":

        label_encoder,

    "all_features":

        all_features,

    "selected_features":

        selected_features,

    "pso_binary_solution":

        global_best_position.tolist(),

    "pso_best_fitness":

        float(
            global_best_fitness
        ),

    "pso_particles":

        NUMBER_OF_PARTICLES,

    "pso_iterations":

        NUMBER_OF_ITERATIONS,

    "target":

        "market_era",

    "best_model_name":

        best_model_name,

    "best_accuracy":

        float(
            best_accuracy
        )

}


with open(

    MODEL_PATH,

    "wb"

) as file:

    pickle.dump(

        model_package,

        file

    )


print(
    "\nPSO best model saved:"
)

print(
    MODEL_PATH
)


# ======================================================================
# 44. ACCURACY GRAPH
# ======================================================================

plt.figure(
    figsize=(11, 6)
)


plt.bar(

    result_df[
        "Model"
    ],

    result_df[
        "Accuracy_Percent"
    ]

)


plt.xlabel(
    "Machine Learning Model"
)


plt.ylabel(
    "Accuracy (%)"
)


plt.title(
    "PSO-Optimized Smart APMC Model Accuracy"
)


plt.ylim(
    0,
    100
)


plt.xticks(

    rotation=25,

    ha="right"

)


for i, value in enumerate(

    result_df[
        "Accuracy_Percent"
    ]

):

    plt.text(

        i,

        value + 1,

        f"{value:.2f}%",

        ha="center"

    )


plt.tight_layout()


plt.savefig(

    ACCURACY_PNG,

    dpi=300,

    bbox_inches="tight"

)


plt.savefig(

    ACCURACY_SVG,

    format="svg",

    bbox_inches="tight"

)


plt.show()

plt.close()


# ======================================================================
# 45. MODEL COMPARISON GRAPH
# ======================================================================

comparison_df = (

    result_df[

        [

            "Model",

            "Accuracy_Percent",

            "Precision_Percent",

            "Recall_Percent",

            "F1_Percent"

        ]

    ]

    .set_index(
        "Model"
    )

)


comparison_df.plot(

    kind="bar",

    figsize=(13, 7)

)


plt.title(
    "PSO-Optimized Model Performance Comparison"
)


plt.xlabel(
    "Machine Learning Model"
)


plt.ylabel(
    "Performance (%)"
)


plt.ylim(
    0,
    100
)


plt.xticks(

    rotation=25,

    ha="right"

)


plt.legend([

    "Accuracy",

    "Precision",

    "Recall",

    "F1 Score"

])


plt.tight_layout()


plt.savefig(

    COMPARISON_PNG,

    dpi=300,

    bbox_inches="tight"

)


plt.savefig(

    COMPARISON_SVG,

    format="svg",

    bbox_inches="tight"

)


plt.show()

plt.close()


# ======================================================================
# 46. HEATMAP DATA
# ======================================================================

heatmap_df = data.copy()


heatmap_df[
    "market_type_code"
] = (

    heatmap_df[
        "market_type"
    ]

    .map({

        "Main Market": 0,

        "Sub Market": 1

    })

)


heatmap_df[
    "market_era_code"
] = (

    heatmap_df[
        "market_era"
    ]

    .map({

        "Legacy (<=1990)": 0,

        "Established (1991-2005)": 1,

        "Modern (2006+)": 2

    })

)


heatmap_columns = [

    "establishment_year",

    "parent_main_year",

    "market_name_length",

    "parent_name_length",

    "area_name_length",

    "market_type_code",

    "market_era_code"

]


heatmap_numeric = (

    heatmap_df[
        heatmap_columns
    ]

    .apply(

        pd.to_numeric,

        errors="coerce"

    )

)


correlation_matrix = (
    heatmap_numeric.corr()
)


# ======================================================================
# 47. HEATMAP
# ======================================================================

plt.figure(
    figsize=(11, 8)
)


image = plt.imshow(

    correlation_matrix,

    cmap="coolwarm",

    aspect="auto",

    vmin=-1,

    vmax=1

)


plt.colorbar(

    image,

    label="Correlation"

)


plt.xticks(

    range(
        len(
            correlation_matrix.columns
        )
    ),

    correlation_matrix.columns,

    rotation=45,

    ha="right"

)


plt.yticks(

    range(
        len(
            correlation_matrix.columns
        )
    ),

    correlation_matrix.columns

)


for i in range(
    correlation_matrix.shape[0]
):

    for j in range(
        correlation_matrix.shape[1]
    ):

        value = (

            correlation_matrix
            .iloc[
                i,
                j
            ]

        )


        if not np.isnan(value):

            plt.text(

                j,

                i,

                f"{value:.2f}",

                ha="center",

                va="center",

                fontsize=9

            )


plt.title(
    "PSO Smart APMC Correlation Heatmap"
)


plt.tight_layout()


plt.savefig(

    HEATMAP_PNG,

    dpi=300,

    bbox_inches="tight"

)


plt.savefig(

    HEATMAP_SVG,

    format="svg",

    bbox_inches="tight"

)


plt.show()

plt.close()


# ======================================================================
# 48. BEST MODEL RESULT GRAPH
# ======================================================================

best_row = result_df.iloc[0]


metric_names = [

    "Accuracy",

    "Precision",

    "Recall",

    "F1 Score"

]


metric_values = [

    best_row[
        "Accuracy_Percent"
    ],

    best_row[
        "Precision_Percent"
    ],

    best_row[
        "Recall_Percent"
    ],

    best_row[
        "F1_Percent"
    ]

]


plt.figure(
    figsize=(9, 6)
)


plt.bar(

    metric_names,

    metric_values

)


plt.xlabel(
    "Evaluation Metric"
)


plt.ylabel(
    "Score (%)"
)


plt.title(

    f"PSO Best Model Performance - "
    f"{best_model_name}"

)


plt.ylim(
    0,
    100
)


for i, value in enumerate(
    metric_values
):

    plt.text(

        i,

        value + 1,

        f"{value:.2f}%",

        ha="center"

    )


plt.tight_layout()


plt.savefig(

    RESULT_GRAPH_PNG,

    dpi=300,

    bbox_inches="tight"

)


plt.savefig(

    RESULT_GRAPH_SVG,

    format="svg",

    bbox_inches="tight"

)


plt.show()

plt.close()


# ======================================================================
# 49. BEST MODEL PREDICTION
# ======================================================================

best_predictions = best_model.predict(

    X_test[
        selected_features
    ]

)


actual_labels = (

    label_encoder
    .inverse_transform(
        y_test
    )

)


predicted_labels = (

    label_encoder
    .inverse_transform(
        best_predictions
    )

)


# ======================================================================
# 50. PREDICTION DATAFRAME
# ======================================================================

prediction_df = pd.DataFrame({

    "Record_Number":

        np.arange(

            1,

            len(
                y_test
            ) + 1

        ),

    "Actual":

        actual_labels,

    "Predicted":

        predicted_labels

})


prediction_df[
    "Correct_Prediction"
] = np.where(

    prediction_df[
        "Actual"
    ]

    ==

    prediction_df[
        "Predicted"
    ],

    "Yes",

    "No"

)


# ======================================================================
# 51. PREDICTION CONFIDENCE
# ======================================================================

if hasattr(

    best_model,

    "predict_proba"

):

    probabilities = best_model.predict_proba(

        X_test[
            selected_features
        ]

    )


    prediction_df[
        "Confidence_Percent"
    ] = (

        probabilities.max(
            axis=1
        )

        * 100

    )


else:

    prediction_df[
        "Confidence_Percent"
    ] = np.nan


prediction_df.to_csv(

    PREDICTION_CSV_PATH,

    index=False

)


print(
    "\nPSO predictions saved:"
)

print(
    PREDICTION_CSV_PATH
)


print(
    "\nSample predictions:"
)


print(

    prediction_df.head(
        15
    )

)


# ======================================================================
# 52. ACTUAL VS PREDICTED GRAPH
# ======================================================================

actual_counts = (

    prediction_df[
        "Actual"
    ]

    .value_counts()

)


predicted_counts = (

    prediction_df[
        "Predicted"
    ]

    .value_counts()

)


classes = list(
    label_encoder.classes_
)


actual_values = [

    actual_counts.get(
        class_name,
        0
    )

    for class_name in classes

]


predicted_values = [

    predicted_counts.get(
        class_name,
        0
    )

    for class_name in classes

]


x = np.arange(
    len(classes)
)


width = 0.35


plt.figure(
    figsize=(11, 7)
)


plt.bar(

    x - width / 2,

    actual_values,

    width,

    label="Actual"

)


plt.bar(

    x + width / 2,

    predicted_values,

    width,

    label="Predicted"

)


plt.xlabel(
    "Market Era"
)


plt.ylabel(
    "Number of Records"
)


plt.title(

    f"PSO Actual vs Predicted - "
    f"{best_model_name}"

)


plt.xticks(

    x,

    classes,

    rotation=20,

    ha="right"

)


plt.legend()


plt.tight_layout()


plt.savefig(

    PREDICTION_GRAPH_PNG,

    dpi=300,

    bbox_inches="tight"

)


plt.savefig(

    PREDICTION_GRAPH_SVG,

    format="svg",

    bbox_inches="tight"

)


plt.show()

plt.close()


# ======================================================================
# 53. CONFUSION MATRIX
# ======================================================================

cm = confusion_matrix(

    y_test,

    best_predictions

)


plt.figure(
    figsize=(9, 7)
)


cm_image = plt.imshow(

    cm,

    cmap="Blues"

)


plt.colorbar(
    cm_image
)


plt.xticks(

    range(
        len(classes)
    ),

    classes,

    rotation=25,

    ha="right"

)


plt.yticks(

    range(
        len(classes)
    ),

    classes

)


for i in range(
    cm.shape[0]
):

    for j in range(
        cm.shape[1]
    ):

        plt.text(

            j,

            i,

            cm[i, j],

            ha="center",

            va="center"

        )


plt.xlabel(
    "Predicted Class"
)


plt.ylabel(
    "Actual Class"
)


plt.title(

    f"PSO Confusion Matrix - "
    f"{best_model_name}"

)


plt.tight_layout()


plt.savefig(

    CONFUSION_PNG,

    dpi=300,

    bbox_inches="tight"

)


plt.savefig(

    CONFUSION_SVG,

    format="svg",

    bbox_inches="tight"

)


plt.show()

plt.close()


# ======================================================================
# 54. VERIFY ALL GENERATED FILES
# ======================================================================

generated_files = [

    CLEANED_DATA_PATH,

    SELECTED_FEATURES_PATH,

    FITNESS_HISTORY_PATH,

    RESULT_CSV_PATH,

    PREDICTION_CSV_PATH,

    MODEL_PATH,

    ACCURACY_PNG,

    ACCURACY_SVG,

    COMPARISON_PNG,

    COMPARISON_SVG,

    HEATMAP_PNG,

    HEATMAP_SVG,

    RESULT_GRAPH_PNG,

    RESULT_GRAPH_SVG,

    PREDICTION_GRAPH_PNG,

    PREDICTION_GRAPH_SVG,

    CONFUSION_PNG,

    CONFUSION_SVG,

    CONVERGENCE_PNG,

    CONVERGENCE_SVG,

    FEATURE_GRAPH_PNG,

    FEATURE_GRAPH_SVG

]


print(
    "\n"
    + "=" * 80
)


print(
    "PSO GENERATED FILES"
)


print(
    "=" * 80
)


for number, file_path in enumerate(

    generated_files,

    start=1

):

    if os.path.exists(
        file_path
    ):

        size_kb = (

            os.path.getsize(
                file_path
            )

            / 1024

        )


        print(

            f"{number}. "

            f"{os.path.basename(file_path)}"

            " | "

            f"{size_kb:.2f} KB"

        )


    else:

        print(

            f"{number}. "

            f"{os.path.basename(file_path)}"

            " | FAILED"

        )


# ======================================================================
# 55. FINAL SUMMARY
# ======================================================================

print(
    "\n"
    + "=" * 80
)


print(
    "PSO SMART APMC MARKET INTELLIGENCE COMPLETED"
)


print(
    "=" * 80
)


print(
    "\nTotal cleaned records:",
    len(data)
)


print(
    "\nOriginal number of features:",
    len(all_features)
)


print(
    "PSO selected features:",
    len(selected_features)
)


print(
    "\nSelected feature names:"
)


for feature in selected_features:

    print(
        " -",
        feature
    )


print(
    "\nPSO best fitness:",
    round(
        global_best_fitness,
        6
    )
)


print(
    "\nBest classification model:",
    best_model_name
)


print(

    "Best testing accuracy:",

    f"{best_accuracy * 100:.2f}%"

)


print(
    "\nAll PSO files saved inside:"
)


print(
    BASE_DIR
)


print(
    "\nDONE!"
)
```
