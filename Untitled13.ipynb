```python
import os
import json
import pickle
import yaml
import pandas as pd
import numpy as np


# ============================================================
# 1. FILE PATHS
# ============================================================

BASE_DIR = r"C:\Users\sagni\Downloads\Smart APMC Market Intelligence"

CSV_PATH = os.path.join(BASE_DIR, "List_Of_APMC.csv")

H5_PATH = os.path.join(BASE_DIR, "apmc_data.h5")
PKL_PATH = os.path.join(BASE_DIR, "apmc_data.pkl")
YAML_PATH = os.path.join(BASE_DIR, "apmc_data.yaml")
JSON_PATH = os.path.join(BASE_DIR, "apmc_data.json")


# ============================================================
# 2. CREATE DIRECTORY IF IT DOES NOT EXIST
# ============================================================

os.makedirs(BASE_DIR, exist_ok=True)

print("=" * 70)
print("SMART APMC MARKET INTELLIGENCE - DATA CONVERSION")
print("=" * 70)

print("\nInput CSV:")
print(CSV_PATH)


# ============================================================
# 3. CHECK CSV EXISTS
# ============================================================

if not os.path.exists(CSV_PATH):
    raise FileNotFoundError(
        f"\nCSV file not found:\n{CSV_PATH}\n"
        "Please check the filename and folder path."
    )


# ============================================================
# 4. LOAD CSV WITH MULTIPLE ENCODING OPTIONS
# ============================================================

def load_csv(file_path):
    """
    Attempts to read CSV using several common encodings.
    """

    encodings = [
        "utf-8",
        "utf-8-sig",
        "cp1252",
        "latin1",
        "ISO-8859-1"
    ]

    for encoding in encodings:

        try:

            df = pd.read_csv(
                file_path,
                encoding=encoding,
                low_memory=False
            )

            print(f"\nCSV loaded successfully using encoding: {encoding}")

            return df

        except UnicodeDecodeError:

            print(f"Encoding failed: {encoding}")

        except Exception as e:

            print(f"Problem while trying {encoding}: {e}")

    raise Exception("Could not read the CSV using available encodings.")


df = load_csv(CSV_PATH)


# ============================================================
# 5. DISPLAY ORIGINAL INFORMATION
# ============================================================

print("\n" + "=" * 70)
print("ORIGINAL DATASET INFORMATION")
print("=" * 70)

print(f"Rows    : {df.shape[0]}")
print(f"Columns : {df.shape[1]}")

print("\nOriginal column names:")

for i, col in enumerate(df.columns, 1):
    print(f"{i}. {col}")


# ============================================================
# 6. CLEAN COLUMN NAMES
# ============================================================

def clean_column_name(column):

    column = str(column).strip()

    column = column.lower()

    column = column.replace(" ", "_")

    column = column.replace("-", "_")

    column = column.replace("/", "_")

    column = column.replace("\\", "_")

    column = column.replace(".", "_")

    column = column.replace("(", "")

    column = column.replace(")", "")

    # Remove multiple underscores
    while "__" in column:
        column = column.replace("__", "_")

    return column


df.columns = [
    clean_column_name(column)
    for column in df.columns
]


# ============================================================
# 7. REMOVE COMPLETELY EMPTY ROWS/COLUMNS
# ============================================================

original_rows = len(df)

df.dropna(
    axis=0,
    how="all",
    inplace=True
)

df.dropna(
    axis=1,
    how="all",
    inplace=True
)

removed_rows = original_rows - len(df)

print(f"\nCompletely empty rows removed: {removed_rows}")


# ============================================================
# 8. REMOVE DUPLICATE ROWS
# ============================================================

duplicate_count = df.duplicated().sum()

df.drop_duplicates(
    inplace=True
)

df.reset_index(
    drop=True,
    inplace=True
)

print(f"Duplicate rows removed: {duplicate_count}")


# ============================================================
# 9. CLEAN TEXT VALUES
# ============================================================

def clean_text(value):

    if pd.isna(value):
        return None

    if isinstance(value, str):

        value = value.strip()

        # Remove unnecessary repeated spaces
        value = " ".join(value.split())

        if value == "":
            return None

    return value


for column in df.columns:

    if df[column].dtype == "object":

        df[column] = df[column].apply(clean_text)


# ============================================================
# 10. REPLACE SPECIAL NUMPY VALUES
# ============================================================

df = df.replace(
    {
        np.nan: None,
        np.inf: None,
        -np.inf: None
    }
)


# ============================================================
# 11. ADD UNIQUE RECORD ID
# ============================================================

if "record_id" not in df.columns:

    df.insert(
        0,
        "record_id",
        range(1, len(df) + 1)
    )


# ============================================================
# 12. DATASET SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("CLEANED DATASET")
print("=" * 70)

print(f"Rows    : {df.shape[0]}")
print(f"Columns : {df.shape[1]}")

print("\nCleaned columns:")

for i, column in enumerate(df.columns, 1):
    print(f"{i}. {column}")


# ============================================================
# 13. CREATE METADATA
# ============================================================

metadata = {

    "project_name":
        "Smart APMC Market Intelligence",

    "dataset_name":
        "List Of APMC",

    "source_file":
        CSV_PATH,

    "number_of_records":
        int(df.shape[0]),

    "number_of_columns":
        int(df.shape[1]),

    "columns":
        list(df.columns),

    "data_types":
        {
            col: str(dtype)
            for col, dtype in df.dtypes.items()
        },

    "missing_values":
        {
            col: int(df[col].isna().sum())
            for col in df.columns
        }
}


# ============================================================
# 14. SAVE PKL FILE
# ============================================================

with open(PKL_PATH, "wb") as file:

    pickle.dump(
        {
            "data": df,
            "metadata": metadata
        },
        file
    )

print("\nPKL generated successfully:")
print(PKL_PATH)


# ============================================================
# 15. SAVE H5 FILE
# ============================================================

# HDF5 works best when unsupported mixed objects are converted.
df_h5 = df.copy()

for column in df_h5.columns:

    if df_h5[column].dtype == "object":

        df_h5[column] = df_h5[column].fillna("").astype(str)


try:

    with pd.HDFStore(
        H5_PATH,
        mode="w"
    ) as store:

        # Main dataset
        store.put(
            "apmc_data",
            df_h5,
            format="table"
        )

        # Metadata
        metadata_df = pd.DataFrame(
            {
                "key": list(metadata.keys()),
                "value": [
                    json.dumps(value, ensure_ascii=False)
                    if isinstance(value, (dict, list))
                    else str(value)
                    for value in metadata.values()
                ]
            }
        )

        store.put(
            "metadata",
            metadata_df,
            format="table"
        )

    print("\nH5 generated successfully:")
    print(H5_PATH)

except ImportError:

    print(
        "\nERROR: PyTables is required for H5.\n"
        "Install it using:\n"
        "pip install tables"
    )


# ============================================================
# 16. PREPARE JSON/YAML SAFE DATA
# ============================================================

records = df.to_dict(
    orient="records"
)


def make_serializable(value):

    # None
    if value is None:
        return None

    # NumPy integer
    if isinstance(value, np.integer):
        return int(value)

    # NumPy float
    if isinstance(value, np.floating):

        if np.isnan(value):
            return None

        return float(value)

    # NumPy bool
    if isinstance(value, np.bool_):
        return bool(value)

    # Pandas timestamp
    if isinstance(value, pd.Timestamp):
        return value.isoformat()

    # Python float NaN
    if isinstance(value, float):

        if np.isnan(value):
            return None

    return value


clean_records = []

for record in records:

    clean_record = {}

    for key, value in record.items():

        clean_record[key] = make_serializable(value)

    clean_records.append(clean_record)


# ============================================================
# 17. SAVE JSON FILE
# ============================================================

json_output = {

    "project": "Smart APMC Market Intelligence",

    "metadata": metadata,

    "apmc_records": clean_records
}


with open(
    JSON_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        json_output,
        file,
        ensure_ascii=False,
        indent=4,
        default=str
    )


print("\nJSON generated successfully:")
print(JSON_PATH)


# ============================================================
# 18. SAVE YAML FILE
# ============================================================

yaml_output = {

    "project":
        "Smart APMC Market Intelligence",

    "metadata":
        metadata,

    "apmc_records":
        clean_records
}


with open(
    YAML_PATH,
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        yaml_output,
        file,
        allow_unicode=True,
        sort_keys=False,
        default_flow_style=False
    )


print("\nYAML generated successfully:")
print(YAML_PATH)


# ============================================================
# 19. VERIFY OUTPUT FILES
# ============================================================

print("\n" + "=" * 70)
print("FILE VERIFICATION")
print("=" * 70)

generated_files = [
    PKL_PATH,
    H5_PATH,
    YAML_PATH,
    JSON_PATH
]


for file_path in generated_files:

    if os.path.exists(file_path):

        size_kb = os.path.getsize(file_path) / 1024

        print(
            f"[OK] {os.path.basename(file_path)} "
            f"({size_kb:.2f} KB)"
        )

    else:

        print(
            f"[FAILED] {os.path.basename(file_path)}"
        )


# ============================================================
# 20. DISPLAY SAMPLE DATA
# ============================================================

print("\n" + "=" * 70)
print("FIRST 5 CLEANED RECORDS")
print("=" * 70)

print(df.head())


# ============================================================
# 21. FINAL MESSAGE
# ============================================================

print("\n" + "=" * 70)
print("PROCESS COMPLETED")
print("=" * 70)

print(
    "\nFiles saved inside:\n"
    + BASE_DIR
)

print(
    "\nGenerated files:"
    "\n1. apmc_data.h5"
    "\n2. apmc_data.pkl"
    "\n3. apmc_data.yaml"
    "\n4. apmc_data.json"
)
```
